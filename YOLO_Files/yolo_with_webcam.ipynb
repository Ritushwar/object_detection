## Working with models

```python
# importing model weights
model = YOLO("/home/ritu/object_detection/YOLO_WEIGHTS/yolov8n.pt")
```

```python
# for total class
total_class = []
with open("/home/ritu/object_detection/classes.csv", mode="r") as file:
    reader = csv.reader(file)
    next(reader)   # skipping heading
    for row in reader:
        total_class.append(row[0])
```

```python
# campture the frame and detect
cap = cv2.VideoCapture(0)
cap.set(3, 640)           # setting width
cap.set(4, 480)           # setting height

while True:
    success, img = cap.read()
    results = model(img, stream=True)
    for r in results:
        boxes = r.boxes
        for box in boxes:
            #x1, y1, x2, y2 = box.xyxy[0]
            #x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
            #print(x1, y1, x2, y2)
            #cv2.rectangle(img, (x1, y1), (x2, y2), (255,0, 255), 3)   # image, (x1, y1), (x2, y2), (color in rgb), width


            # if you want to use cvzone for fancy recatnge
            x1, y1 , x2 , y2 = box.xyxy[0]
            x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
            w, h = x2-x1, y2-y1
            cvzone.cornerRect(img, (x1, y1, w, h))

            # confidence
            conf = math.ceil(box.conf[0]*100)/100
            #print(conf)
            #cvzone.putTextRect(img, f'{conf}', (max(0,x1),max(30,y1)))
            # class
            clls = int(box.cls[0])
            cvzone.putTextRect(img, f'{conf}{total_class[clls]}', (max(0,x1),max(30,y1)), scale=0.5, thickness=1)

    cv2.imshow("Images", img)
    key = cv2.waitKey(1) & 0xFF   # masking only last 8 bit
    if key == ord('q'):   # for key q
        break
    elif key == 27:       # for esc
        break

cap.release()
cv2.destroyAllWindows()
```
