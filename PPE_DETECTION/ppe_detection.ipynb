# PPE DETECTION

## IMPORTING REQUIRED LIBRARIES

```python
from ultralytics import YOLO
import cv2
import cvzone
import math
import csv
import torch
import numpy as np
```

## IMPORTING MODEL

```python
model = YOLO("/home/ritu/object_detection/PPE_DETECTION/model.pt")
```

## IDENTIFYING TOTAL CLASSES

```python
total_classes = []
classes = model.names
for clss in classes:
    total_classes.append(classes[clss])
print(len(total_classes))
```

```python
total_classes
```

## PASSING IT TO THE MODEL

```python
cap = cv2.VideoCapture("/home/ritu/object_detection/PPE_DETECTION/test2.mp4")
frame_width = 640
frame_height = 480
while True:
    success, img = cap.read()
    if not success:
        break
    img = cv2.resize(img, (frame_width,frame_height))
    results = model(img, stream=True)
    for r in results:
        boxes = r.boxes
        for box in boxes:
            # Bounding Box
            x1, y1, x2, y2 = box.xyxy[0]
            x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
            conf = math.ceil((box.conf[0] * 100)) / 100
            clss = int(box.cls[0])
            current_class = total_classes[clss]
            w , h = x2 - x1, y2 - y1
            if (current_class=="NO-Safety Vest" or current_class=="NO-Hardhat" or current_class=="NO-Mask" ) and (conf > 0.3):
                cvzone.cornerRect(img, (x1,y1,w,h), colorC=(0,0,255),l=5)
                cvzone.putTextRect(img, f'{current_class}', (max(0,x1),max(0,y1)), scale=0.6, thickness=1)
            if (current_class=="Hardhat" or current_class=="Mask" or current_class=="Hardhat" or current_class=="Safety Vest") and (conf>0.3):
                cvzone.cornerRect(img, (x1,y1,w,h), colorC=(0,255,0),l=5)
                cvzone.putTextRect(img, f'{current_class}', (max(0,x1),max(0,y1)), scale=0.6, thickness=1)
            if current_class=="Person" and conf > 0.3:
                cvzone.cornerRect(img, (x1,y1,w,h), colorC=(255,0,0),l=5)
                cvzone.putTextRect(img, f'{current_class}', (max(0,x1),max(0,y1)), scale=0.6, thickness=1)
    cv2.imshow("Result", img)
    key = cv2.waitKey(1) & 0xFF
    if key == ord('q'):
        break
    elif key == 27:
        break
cap.release()
cv2.destroyAllWindows()
```
